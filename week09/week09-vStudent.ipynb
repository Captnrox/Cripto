## Challenge 3

In the Diffie-Hellman protocol, and given the public parameters p, prime, and g ∈ Zp, each
participant selects a secret number x and sends the other participant g^x mod p. What
would happen if the participants sent each other x^g mod p instead? Give one method Alice
and Bob could use it to agree on a key. Can Eve (a passive attacker) break the system without
finding the secret numbers? Can Eve find the secret numbers?

In the given scheme, two participants are forced to share their x^g mod p values to communicate a secret. Supposing Alice and Bob from the previous exercise, Alice shares A, a^g and Bob shares B, b^g back to Alice. In this case, they calculate the secret by doing A\*B. This, however, is not safe against eavesdropping attacks. Eve listening to the channel, can obtain A and B, and, with it, obtain A*B, therefore listening in to the secret. This means there is no need to find the secret numbers a and b, especially because there is no efficient way to calculate them, both exponent and base problems have similar complexities.
