## Chalenge 6

Use openSSL to generate a key pair where private key is protected with a password.

openssl genrsa 4096

See what happens when you increase/decrease the key size.
Investigate how openSSL converts the passphrase into a cryptography key for encryption/wrapping.


openssl genpkey -algorithm RSA -out private_key.pem -aes256 -pass pass:password -pkeyopt rsa_keygen_bits:4096

The size of the private key file increases with the increase of the key size, a 2048-bit RSA shoud be about 1700 bytes but a 4096-bit RSA is around the double, 3400 bytes. OpenSSÇ uses a key deriation funciton KDF to convert the passphrase into a cryptographic key, which then encrypts the private key file. Tipically PBKDF2 is used. The process invovles hashing the passsphrase multiple ties to produce a strong key, iteratively adding a salt value to enhance security, making brute-force attacks way harder to be done. After deriving the key, OpenSSL uses AES-256 encryption to encrypt the file.All salt values used along hashing and encryption are stored in the header, allowing for decryption of the files and guaranteeing different values even if the same password is used twice in different files.
