## Challenge 2

In a RSA system, the public key of a given user is e = 65, n = 2881. What is the private key of this user?

Having a given e and an n we can determine the value of d, the private key, using the extended version of Euclids algorithm:

65 * x + Φ(2881) * y = 1

2881 = 43*67, Φ(2881) = 42 * 66 =  2772

65 * x + 2772 * y = 1.

```python
def gcdExtended(a, b): 
    # Base Case 
    if a == 0 : 
        return b,0,1
             
    gcd,x1,y1 = gcdExtended(b%a, a) 
     
    # Update x and y using results of recursive 
    # call 
    x = y1 - (b//a) * x1 
    y = x1 
     
    return gcd,x,y 
     
 
# Driver code 
a, b = 2772, 65
g, x, y = gcdExtended(a, b) 
print(g,x,y) 
```

We obtain x = 725, in other words, d or the private key is 725.

## Challenge 4

Suppose Bob uses the RSA cryptosystem with a very large modulus n for which the factorisation cannot be found in a reasonable amount of time. Suppose Alice sends an enciphered
message to Bob containing only her phone number: number^e (mod n). Is this safe?

While by itself there seems no problem in implementation since it seems to follow how the basic RSA implementation is done, the lack of padding and randomness makes it deterministic. In this way, if the attacker knows the secret message is a phone number, all they have to do is build a dictionary with all possible phone numbers using Bob's e and the ppublicly known n. This way they can find the value of the phone number, making it not secure.
