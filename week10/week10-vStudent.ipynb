## Challenge 5
Write python/SageMath programs that for P-1921 and ECDSA

(a) Generates a pair of private/public keys.

(b) Sign a text using a private key.

(c) Verifies a signature using a public key.

```python
from fastecdsa import keys, curve, ecdsa
import hashlib


curve_p192k1 = curve.secp192k1

private_key = keys.gen_private_key(curve_p192k1)
public_key = keys.get_public_key(private_key, curve_p192k1)

print(f"Private Key: {private_key}")
print(f"Public Key: ({public_key.x}, {public_key.y})")

message = input("Write a message:")
print(f"Message: {message}")
enc_message = message.encode('utf-8')

hashed_message = hashlib.sha256(enc_message).digest()
hashed_message_int = int.from_bytes(hashed_message, 'big')
signature = ecdsa.sign(hashed_message, private_key, curve_p192k1)

print(f"Signature: {signature}")

is_valid = ecdsa.verify(signature, hashed_message, public_key, curve_p192k1)

print(f"Signature valid: {is_valid}")

```
